```python
import pandas as pd
from imblearn.over_sampling import RandomOverSampler
import seaborn as sns
import matplotlib.pyplot as plt

import src.utils as utils

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
```

## Main

```python
X_train_prep = utils.deserialize_data("data/processed/X_train_prep.pkl")
y_train_prep = utils.deserialize_data("data/processed/y_train_prep.pkl")
```

```python
ros = RandomOverSampler(random_state=42)
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(data=pd.DataFrame(y_train_prep, columns=['loan_status']), x='loan_status', hue='loan_status')
plt.title('Distribusi Label Sebelum Oversampling')
plt.show()
```

```python
X_train_ros, y_train_ros = ros.fit_resample(X_train_prep, y_train_prep) # type: ignore
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(data=pd.DataFrame(y_train_ros, columns=['loan_status']), x='loan_status', hue='loan_status')
plt.title('Distribusi Label Setelah Oversampling')
plt.show()
```

```python
utils.serialize_data(X_train_ros, "data/processed/X_train_ros.pkl")
utils.serialize_data(y_train_ros, "data/processed/y_train_ros.pkl")
```
